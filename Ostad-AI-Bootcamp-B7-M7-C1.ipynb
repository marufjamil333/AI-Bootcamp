```python
import pandas as pd

df = pd.read_csv("data.csv")

# JSON
df.to_json("data.json", orient="records", indent=2)
df_json = pd.read_json("data.json")
print(df_json.head())

# Parquet
df.to_parquet("data.parquet", index=False)
df_parquet = pd.read_parquet("data.parquet")
print(df_parquet.head())
```
