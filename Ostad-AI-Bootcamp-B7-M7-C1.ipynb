```python
import pandas as pd
df=pd.read_csv("Titanic-Dataset.csv")
```

```python
df.head()
```

```python
#4. NOT Condition
# select passengers whose age is NOT greater than 30
df[~(df["Age"] > 30)]

# select passengers whose not in first class
df[~(df["Pclass"] == 1)]
```

```python
# 5. Using isin()
# select passengers who embarked from 'C' or 'S'
df[df["Embarked"].isin(["C", "S"])]

# select passengers who does not embarked from 'C' or 'S'
df[~df["Embarked"].isin(["C", "S"])]
```

```python
df.head()
```

```python
#Range filtering
#select passengers whose age is between 20 and 30

df[df["Age"].between(20, 30)]

```

```python
#Missing values filtering
#select row where age is missing
df[df["Age"].isnull()]
```

```python
# select rows where age is NOT missing
df[df["Age"].notnull()]
```

```python
# select rows where name contains 'John'
df[df["Name"].str.contains("John")]

```

```python
# select rows where name starts with 'A'
df[df["Name"].str.startswith("A")]

```

```python
# select rows where name ends with 'son'
df[df["Name"].str.endswith("son")]
```

```python
#9. Query Method (Readable Alternative)

df.query("Age > 30 and Fare > 50 ")
```

```python
#Conditional Assignments
 #create/update  'catagory ' column as 'child ' where age <18
df.loc[df["Age"] < 18, "Category"] = "Child"

```

```python
df.loc[df["Age"] >= 18, "Category"] = "Adult"
```

```python
# 11. Counting with Condition
# count number of passengers whose age > 30
(df["Age"] > 30).sum()
```

```python
# ❌ WRONG (will throw error)
df[df["Age"] > 30 and df["Fare"] > 50]
```

```python
# ✅ CORRECT (use & and parentheses)
df[(df["Age"] > 30) & (df["Fare"] > 50)]
```

```python
df.groupby(["Pclass", "Sex"])["Age"].mean()
```

```python
# total fare per gender
df.groupby("Sex")["Fare"].sum()

```

```python
# count of records per gender
df.groupby("Sex")["Age"].count()
```

```python
#MUltiple aggregations
df.groupby("Sex").agg({"Fare": "sum", "Age": "count"})
```

```python
#groupby with multiple columns
df.groupby(["Pclass", "Sex"]).agg({"Fare": "sum", "Age": "count"})
```

```python
#reset index very common
df.groupby("Sex")["Age"].mean().reset_index()
```

```python
df.groupby("Sex").agg( avg_age=("Age", "mean"),
 max_fare=("Fare","max"))
```

```python
# keep groups where average age > 30
df.groupby("Sex").filter(lambda x: x["Age"].mean() > 30)
```
